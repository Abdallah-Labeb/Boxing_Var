```python
# 1. إجبار الكولاب على قراءة الملفات الجديدة
import sys
project_path = '/content/drive/MyDrive/cnndataset'
if project_path not in sys.path:
    sys.path.append(project_path)

# 2. استدعاء كل الملفات اللي عملناها
import torch
from torch.utils.data import DataLoader, random_split
from dataset import BoxingVideoDataset, BoxingHitDataset
from action_model import ActionTransformer
from hit_model import HitNet
from trainer import Trainer

# إعداد الجهاز (GPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using Device: {device}")

# ==========================================
# أولاً: تدريب موديل ActionNet (الفيديو)
# ==========================================
print("\n🎥 1. Training ActionNet (Transformer)...")

# تحميل الداتا
video_root = '/content/drive/MyDrive/cnndataset/Boxing_clip'
# بنختار seq_len=16 عشان ده اللي الموديل متصمم عليه
video_dataset = BoxingVideoDataset(root_dir=video_root, seq_len=16)

# تقسيم الداتا (80% تدريب - 20% اختبار)
train_size = int(0.8 * len(video_dataset))
val_size = len(video_dataset) - train_size
train_ds, val_ds = random_split(video_dataset, [train_size, val_size])

train_loader = DataLoader(train_ds, batch_size=4, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=4, shuffle=False)

# تعريف الموديل
action_model = ActionTransformer(num_classes=3, num_frames=16)

# بدء التدريب (هنجرب 10 دورات)
trainer_action = Trainer(action_model, train_loader, val_loader, device=device, model_name="ActionNet")
trainer_action.fit(epochs=10)


# ==========================================
# ثانياً: تدريب موديل HitNet (الصور)
# ==========================================
print("\n🥊 2. Training HitNet (Attention CNN)...")

image_root = '/content/drive/MyDrive/cnndataset/punch_imgs'
image_dataset = BoxingHitDataset(root_dir=image_root)

train_size = int(0.8 * len(image_dataset))
val_size = len(image_dataset) - train_size
train_ds_img, val_ds_img = random_split(image_dataset, [train_size, val_size])

train_loader_img = DataLoader(train_ds_img, batch_size=8, shuffle=True)
val_loader_img = DataLoader(val_ds_img, batch_size=8, shuffle=False)

# تعريف الموديل
hit_model = HitNet()

# بدء التدريب
trainer_hit = Trainer(hit_model, train_loader_img, val_loader_img, device=device, model_name="HitNet")
trainer_hit.fit(epochs=10)
```
